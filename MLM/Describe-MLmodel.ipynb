# Describe the exported model with STAC MLM

Run training and export first. The model checksum is verified against its adjacent inference contract before publishing metadata. Band order, normalization, shape and class order come from the exported model contract.


```python
from pathlib import Path
import os
from mlm.describe import describe
model = Path(os.environ.get("EOAP_MODEL", "runs/labs/app/model.onnx")).resolve()
assert model.exists(), "Run the training and export lessons first"
```

```python
item = describe(model, model.parent / "model-stac.json")
assert item.properties["mlm:input"][0]["bands"]
assert item.assets["model"].extra_fields["mlm:artifact_type"] == "onnx"
item.to_dict()
```
